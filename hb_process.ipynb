```python
# merge the two dataframes
fgsdf = pd.concat([fgsdf1, fgsdf2], ignore_index=True)
show_schema(fgsdf)
```

```python
print( 'Duplicate:\n', fgsdf[fgsdf['#Chinese'] == "Duplicate" ].count())
print( 'NaN:\n', fgsdf[fgsdf['#English'] == None ].count())
print( 'Removed:\n', fgsdf[fgsdf['#Chinese'] == "Removed" ].count())
print( 'Lost:\n', fgsdf[fgsdf['#Chinese'] == "Lost" ].count())



```
