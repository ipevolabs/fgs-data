```python
fgsdf_resolved = pd.read_csv('different_english_resolved.tsv', sep='\t')
print('columns:', fgsdf_resolved.columns)
print('nrows', fgsdf_resolved.count())

```

```python
def fgsdf_resolute( fgsdf: pd.DataFrame, colnm:str):
	fgsdf[colnm] = fgsdf[f'{colnm}_df1'].fillna(fgsdf[f'{colnm}_df2'])
```

```python
#for each row, take the first non-empty value of the column "#English"
fgsdf_resolute(fgsdf_resolved, '#English')
```

```python
#remove the columns "#English_df1" and "#English_df2"
fgsdf_resolved.drop(columns=['#English_df1', '#English_df2'], inplace=True)
#dump the first 10 rows
fgsdf_resolved.head(10)
```

```python
#check if any row has empty value in the column "#English"
fgsdf_resolved[fgsdf_resolved['#English'].isnull()]
```

```python
#for those duplicated rows, just keep the first one
fgsdf_resolved_clean = fgsdf_resolved.drop_duplicates(subset='#Chinese', keep='first')
fgsdf_resolved_clean.count()
```

```python
#duplicate wfgsdf to wfgsdf_original
wfgsdf1 = wfgsdf.copy()
#for each row in fgsdf_resolved_clean, if wfgsdf has the same row on the Chinese column, in-place remove that row 
for index, row in fgsdf_resolved_clean.iterrows():
	wfgsdf1.drop( wfgsdf1[wfgsdf1['#Chinese'] == row['#Chinese']].index, inplace=True)  

# merge wgfsdf1 and fgsdf_resolved_clean
wfgsdf1 = pd.concat([wfgsdf1, fgsdf_resolved_clean], ignore_index=True)
```
